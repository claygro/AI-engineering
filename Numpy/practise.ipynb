# Advance operation with Business examples


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("=== Zomato sales analysis===")
print("\n Sales data shape", sales_data.shape)
print("\n sample data for first 3 resturants:\n", sales_data[:, 1:]) #first : means all row and 1: means get all column from 1 index
```

```python
# total sales
print(np.sum(sales_data, axis=0)) # 0 means column wise and 1 means row wise

yearly_total = np.sum(sales_data[:, 1:], axis=0)
print("Yearly total: ", yearly_total)


# Minimun sales per resturant
min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)

# Maximun sales per year
max_sales = np.max(sales_data[:, 1:], axis=0)
print(max_sales)

# Average sales per resturants
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales accross all the restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])
print("Vector addition: ", vector1 + vector2)
print("\n Multiplication vector: ", vector1 * vector2)
print("\n Dot product: ", np.dot(vector1, vector2))
print("\n Angle: ", np.arccos(np.dot(vector1, vector2)/ (np.linalg.norm(vector1) * np.linalg.norm(vector2)))) # cosA = (a.b)/mag(a).mag(b)
```
