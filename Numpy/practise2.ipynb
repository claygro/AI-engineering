# Analysing 50 students

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
std_data = np.genfromtxt("student_performance_50.csv", delimiter = ",", skip_header=0, names=True, dtype=None, encoding="utf-8")
print(std_data)
```

## Extract only student id, math, science, english, computer

```python
numeric_data = np.column_stack((std_data["student_id"], std_data["math"], std_data["science"], std_data["english"], std_data["computer"]))
print(numeric_data)
```

## Performing calculation

```python
print("Mean: ", np.mean(numeric_data[:, 1:], axis=0))
print("Median: ", np.median(numeric_data[:, 1:], axis=0))
print("Average: ", np.average(numeric_data[:, 1:], axis=0))
print("Average: ", np.average(numeric_data[:, 1:], axis=0))
```

### Filtering the data

```python
filtered_data = np.where(numeric_data>80, numeric_data, 0)
print("Filtered data: ", filtered_data)
```
