# Dataframe

When we combine multiple series together then it is called dataframe

```python
import numpy as np
import pandas as pd
```

### Dataframe using dictionary

```python
data = {
    "name": ["John", "Research", "Ram", "Sita"],
    "age": [25, 22, 30, 28],
    "city": ["New York", "Bharatpur", "Kathmandu", "Pokhara"],
    "salary": [50000, 45000, 60000, 55000]
}
```

```python
pd.DataFrame(data)
```

### Dataframe using list

```python
data = [
    ["John", "New York", 25, 60000],
    ["Research", "Bharatpur", 22, 45000],
    ["Ram", "Kathmandu", 30, 60000],
    ["Sita", "Pokhara", 28, 55000]
]
```

```python
columns =  ["Name", "City", "Age", "Salary"]
df2 = pd.DataFrame(data, columns=columns)
```

### Selection and indexing of column

 - For selection and indexing, we dont use index like [0], [1] etc we use column or header name like ["Name"], ["City"] or for selecting multiple column we use array of aaray like [["Name", "Column"]]

```python
df2["City"]

df2[["Name", "City"]]
```

### Adding new column

```python
df2["Designation"] = ["Doctor", "engineer", "Doctor", "engineer"]
print(df2)
```

### Remove column

```python
df2.drop(["Age", "Salary"], axis = 1)
```

### Selection of row

```python
df2.loc[0] # loc is location which is used to selecting data from row
```

### Selecting subsets of rows and columns

```python
print(df2)
df2.loc[[2, 3]][["Name", "City"]]
```

### Conditional selection


```python
print(df2)
print("\n", df2[df2["Age"] > 22])

print("\n", df2[(df2["Age"] > 22) & (df2["City"] == "Kathmandu")])
print("\n", df2[(df2["Age"] > 22) | (df2["City"] == "Kathmandu")])
```
