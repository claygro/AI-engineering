```python
import numpy as np
import pandas as pd
```

```python
data = {
    "A": [1, 2, np.nan, 4, 5],
    "B": [1, 2, 3, 4, 5],
    "C": [1, 2, 3, np.nan, np.nan],
    "D": [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
df
```

### Finding missing data

```python
df.isna() # check whether row contains NaN value or not. If contains, it will true if not, it will false 
```

```python
df.isna().sum()
```

```python
df.isna().any() # check whether any column contains null value 
```

### Drop the row if it contains missing data

```python
# Drop a NaN value. It drop on the basis of row. 
# If any NaN value contains by row, then it will remove that row completely 

df.dropna()
```

```python
# thresh = 3 means don't remove the row if it contains three not NaN value
df.dropna(thresh=2)
```

### Filling the missing data

```python

# it will not do any changes in the original data frame unless we add inplace = true
df.fillna(0) # it is column wise
```

### Filling missing data accoring to column

```python
values = {"A": 0, "B": 100, "C": 300, "D": 400}
df.fillna(value=values) # it is column wise
```
