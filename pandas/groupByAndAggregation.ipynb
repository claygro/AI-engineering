```python
import numpy as np
import pandas as pd
```

```python

data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
```

```python
cat = df.groupby("Category")["Sales"].sum()
cat
```

```python
cat = df.groupby("Store")["Sales"].sum()
cat
```

```python
cat = df.groupby(["Category", "Store"])["Sales"].sum()
cat
```

### Aggregation function

```python
print(df)
print("Mean: \n", df["Sales"].mean())
print("Median: \n", df["Sales"].median())
print("Min: \n", df["Sales"].min())
print("Max: \n", df["Sales"].max())
print("Count: \n", df["Sales"].count())
print("Standard deviation: \n", df["Sales"].std())
print("Variance: \n", df["Sales"].var())
```

```python
df["Sales"].agg(["mean", "median", "min", "max"])
```
