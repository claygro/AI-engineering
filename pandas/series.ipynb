# Series

series is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called index.

```python
import numpy as np
import pandas as pd
```

```python
labels = ["a", "b", "c"]
my_list = [10, 20, 30]
np_arr = np.array([10, 20, 30])
d = {1:10, 2:20, 3:30}
pd.Series(my_list)
```

### With label

```python
pd.Series(my_list, index=labels)
```

### with array


```python
pd.Series(np_arr, index=labels)
```

### with dictionary

```python
pd.Series(d)
```
